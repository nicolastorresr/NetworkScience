# Propagación de la información

Proceso mediante el cual una pieza de información (conocimiento) se difunde y llega a los individuos a través de interacciones.

## Elementos

1. **Remitente(s).** Un remitente o un pequeño conjunto de remitentes inician el proceso de difusión de información.

2. **Receptor(es).** Un receptor o un conjunto de receptores reciben información difundida. Comúnmente, el conjunto de receptores es mucho mayor que el conjunto de remitentes y puede superponerse con el conjunto de remitentes.

3. **Medio.** Este es el medio a través del cual se produce la difusión. Por ejemplo, cuando se difunde un rumor, el medio puede ser la comunicación personal entre individuos.

### Independent Cascade Model (ICM)

* La red se representa mediante un grafo dirigido.
* Las decisiones son binarias: los nodos pueden estar *activos* o *inactivos*.
* Un nodo, una vez activado, tiene una sola oportunidad de activar a sus vecinos con una cierta *probabilidad*.
* La activación es un proceso *progresivo*, donde los nodos cambian de inactivos a activos, pero no al revés.
* El proceso comienza con un conjunto de nodos activos y continúa hasta que ya no sea posible ninguna activación adicional.

```python
import networkx as nx
import random
import matplotlib.pyplot as plt
```

```python
def independent_cascade(G, initial_active_nodes, activation_prob):
    """
    Simulación del modelo de cascada independiente (ICM) en una red dirigida.

    Parámetros:
    - G: Grafo dirigido representando la red.
    - initial_active_nodes: Lista de nodos inicialmente activos.
    - activation_prob: Probabilidad de activación de un vecino.

    Retorna:
    - active_nodes: Conjunto final de nodos activos después de que el proceso se detenga.
    """
    active_nodes = set(initial_active_nodes)  # Nodos activos al principio
    new_active_nodes = set(initial_active_nodes)  # Nodos que se activarán en la iteración actual

    while new_active_nodes:
        next_active_nodes = set()  # Nodos que se activarán en la próxima iteración
        for node in new_active_nodes:
            for neighbor in G.successors(node):
                if neighbor not in active_nodes:  # Solo intentamos activar nodos inactivos
                    if random.random() <= activation_prob:  # Probabilidad de activar al vecino
                        next_active_nodes.add(neighbor)

        active_nodes.update(next_active_nodes)  # Actualizamos los nodos activos
        new_active_nodes = next_active_nodes  # Los nuevos activados se vuelven los actuales en la siguiente iteración

    return active_nodes
```

```python
def plot_graph(G, active_nodes):
    """
    Dibuja el grafo resaltando los nodos activos.
    
    Parámetros:
    - G: Grafo dirigido.
    - active_nodes: Conjunto de nodos activos.
    """
    pos = nx.spring_layout(G)  # Posiciones de los nodos
    plt.figure(figsize=(8, 6))
    
    # Dibujamos nodos inactivos en azul
    inactive_nodes = set(G.nodes()) - active_nodes
    nx.draw_networkx_nodes(G, pos, nodelist=inactive_nodes, node_color='lightblue', node_size=500, alpha=0.8)
    
    # Dibujamos nodos activos en rojo
    nx.draw_networkx_nodes(G, pos, nodelist=active_nodes, node_color='red', node_size=500, alpha=0.8)
    
    # Dibujamos las aristas del grafo
    nx.draw_networkx_edges(G, pos, arrowstyle='->', arrowsize=20, edge_color='gray')
    
    # Dibujamos etiquetas de los nodos
    nx.draw_networkx_labels(G, pos, font_size=12)
    
    plt.title("Simulación del Modelo de Cascada Independiente (ICM)")
    plt.show()
```

```python
# Creación de un grafo dirigido
G = nx.DiGraph()
# Añadimos nodos y aristas (ejemplo de una pequeña red)
G.add_edges_from([(0, 1), (0, 2), (1, 3), (2, 3), (3, 4)])
```

```python
# Simulamos el ICM
active_nodes = independent_cascade(G, initial_active_nodes=[0], activation_prob=0.5)
# Mostramos el grafo con los nodos activos resaltados
plot_graph(G, active_nodes)
print("Nodos activos al final del proceso:", active_nodes)
```

### Innovaciones

Una **innovación** es una idea, práctica u objeto que es percibido como nuevo por un individuo o un grupo de personas que lo potencialmente lo pueden adoptar. Las innovaciones se crean regularmente; sin embargo, no todas las innovaciones se propagan entre las poblaciones.

#### Características de la innovación

Para que una innovación sea adoptada, el individuo que la adopta y la innovación deben tener ciertas cualidades. Las innovaciones deben ser altamente *observables*, deben tener una *ventaja relativa* sobre las prácticas actuales, deben ser *compatibles* con el paradigma sociocultural al que se presentan, deben tener una *vida útil* suficiente para funcionar varias veces y no deben ser altamente *complejas*.

## Epidemias

En los modelos epidémicos, aunque similares a los modelos de difusión de innovaciones, los individuos no deciden si infectarse o no y la infección se considera un proceso natural aleatorio, siempre que el individuo esté expuesto al patógeno.

<div align="center">
    <img src="img/epidemics.png" alt="" width="30%" align="center" />
</div>

### Elementos

- **Patógeno:** la enfermedad que se propaga, publicación/meme que se comparte.
- **Población de huéspedes:** humanos, animales, plantas, entre otros. 
- **Mecanismo de propagación:** respiración, alimentos, actividades sociales, internet, etc.

<div align="center">
    <img src="img/Host-Pathogen-Interaction.png" alt="" width="50%" align="center" />
</div>

### Definiciones

- **Susceptible:** puede potencialmente infectarse por la enfermedad.
- **Infectado:** tiene la posibilidad de infectar a individuos susceptibles.
- **Recuperados (o eliminados):** se trata de personas que se han recuperado de la enfermedad y, por lo tanto, tienen inmunidad total o parcial contra la infección o que murieron a causa de la infección.

<div align="center">
    <img src="img/definitions.png" alt="" width="45%" align="center" />
</div>

### SI Model

En este modelo, las personas pueden estar en dos estados: Susceptibles (S) o Infectadas (I). Una vez que una persona se infecta, permanece infectada para siempre y puede seguir contagiando a los susceptibles con probabilidad $\beta$. Por ejemplo: enfermedades crónicas como el VIH.

### SIS Model

En este modelo, una persona infectada puede recuperarse, pero no desarrolla inmunidad, por lo que puede volver a ser susceptible con probabilidad $\gamma$. Así, los individuos alternan entre los estados de Susceptible e Infectado. Por ejemplo: Resfriado común.

<div align="center">
    <img src="img/SI-SIS.png" alt="" width="30%" align="center" />
</div>

### SIR Model

Aquí, los individuos pueden recuperarse y, tras la recuperación, se vuelven inmunes, es decir, no pueden volver a infectarse. Este modelo tiene tres estados: Susceptible (S), Infectado (I), y Recuperado (R). Por ejemplo, el sarampión.

### SIRS Model

Similar al modelo SIR, pero los individuos pueden perder la inmunidad después de un tiempo y volver a ser susceptibles con probabilidad $\xi$. Ejemplo: Gripe (Influenza).

<div align="center">
    <img src="img/SIR-SIRS.png" alt="" width="50%" align="center" />
</div>

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Funciones para simular cada modelo
def si_model(beta, S0, I0, T):
    S, I = [S0], [I0]
    for t in range(T):
        new_infected = beta * S[-1] * I[-1]
        S.append(S[-1] - new_infected)
        I.append(I[-1] + new_infected)
    return np.array(S), np.array(I)
```

```python
def sis_model(beta, gamma, S0, I0, T):
    S, I = [S0], [I0]
    for t in range(T):
        new_infected = beta * S[-1] * I[-1]
        new_recovered = gamma * I[-1]
        S.append(S[-1] + new_recovered - new_infected)
        I.append(I[-1] + new_infected - new_recovered)
    return np.array(S), np.array(I)
```

```python
def sir_model(beta, gamma, S0, I0, T):
    S, I, R = [S0], [I0], [0]
    for t in range(T):
        new_infected = beta * S[-1] * I[-1]
        new_recovered = gamma * I[-1]
        S.append(S[-1] - new_infected)
        I.append(I[-1] + new_infected - new_recovered)
        R.append(R[-1] + new_recovered)
    return np.array(S), np.array(I), np.array(R)
```

```python
def sirs_model(beta, gamma, xi, S0, I0, T):
    S, I, R = [S0], [I0], [0]
    for t in range(T):
        new_infected = beta * S[-1] * I[-1]
        new_recovered = gamma * I[-1]
        new_susceptible = xi * R[-1]
        S.append(S[-1] - new_infected + new_susceptible)
        I.append(I[-1] + new_infected - new_recovered)
        R.append(R[-1] + new_recovered - new_susceptible)
    return np.array(S), np.array(I), np.array(R)
```

```python
def plot_epidemics(
                    beta = 0.3,    # Tasa de infección
                    gamma = 0.1,   # Tasa de recuperación
                    xi = 0.05,     # Tasa de pérdida de inmunidad (para SIRS)
                    S0 = 0.99,     # Proporción inicial de susceptibles
                    I0 = 0.01,     # Proporción inicial de infectados
                    T = 100,       # Número de pasos de tiempo
                    ):
    # Simulación
    S_si, I_si = si_model(beta, S0, I0, T)
    S_sis, I_sis = sis_model(beta, gamma, S0, I0, T)
    S_sir, I_sir, R_sir = sir_model(beta, gamma, S0, I0, T)
    S_sirs, I_sirs, R_sirs = sirs_model(beta, gamma, xi, S0, I0, T)
    
    # Graficamos los resultados
    fig, axes = plt.subplots(2, 2, figsize=(12, 10))

    # SI Model
    axes[0, 0].plot(S_si, label='S (Susceptibles)')
    axes[0, 0].plot(I_si, label='I (Infectados)')
    axes[0, 0].set_title('SI Model')
    axes[0, 0].set_xlabel('Tiempo')
    axes[0, 0].set_ylabel('Proporción de la población')
    axes[0, 0].legend()

    # SIS Model
    axes[0, 1].plot(S_sis, label='S (Susceptibles)')
    axes[0, 1].plot(I_sis, label='I (Infectados)')
    axes[0, 1].set_title('SIS Model')
    axes[0, 1].set_xlabel('Tiempo')
    axes[0, 1].set_ylabel('Proporción de la población')
    axes[0, 1].legend()

    # SIR Model
    axes[1, 0].plot(S_sir, label='S (Susceptibles)')
    axes[1, 0].plot(I_sir, label='I (Infectados)')
    axes[1, 0].plot(R_sir, label='R (Recuperados)')
    axes[1, 0].set_title('SIR Model')
    axes[1, 0].set_xlabel('Tiempo')
    axes[1, 0].set_ylabel('Proporción de la población')
    axes[1, 0].legend()

    # SIRS Model
    axes[1, 1].plot(S_sirs, label='S (Susceptibles)')
    axes[1, 1].plot(I_sirs, label='I (Infectados)')
    axes[1, 1].plot(R_sirs, label='R (Recuperados)')
    axes[1, 1].set_title('SIRS Model')
    axes[1, 1].set_xlabel('Tiempo')
    axes[1, 1].set_ylabel('Proporción de la población')
    axes[1, 1].legend()

    plt.tight_layout()
    plt.show()
```

```python
plot_epidemics(beta = 0.3, gamma = 0.1, xi = 0.05, S0 = 0.99, I0 = 0.01, T = 100)
```
