```python
import os
def download_profile_images(username_file, output_folder):
    L = instaloader.Instaloader()
    with open(username_file, 'r') as file:
        usernames = file.read().splitlines()

    for username in usernames:
        try:         
            profile = instaloader.Profile.from_username(L.context, username)
            
            response = requests.get(profile.profile_pic_url)

            if response.status_code == 200:
                image_data = response.content
                save_path = f"{output_folder}/{username}.jpg"
                with open(save_path, 'wb') as image_file:
                    image_file.write(image_data)

                print(f"Profile Image from {username} downloaded and saved as {username}.jpg")
            else:
                print(f"Failed to download image from {url}. Status code: {response.status_code}")


            print(f"Downloaded profile image for {username}")
        except Exception as e:
            print(f"Error downloading profile image for {username}: {e}")
```

```python
username_file = "usernames.txt"
output_folder = "profile_images"

download_profile_images(username_file, output_folder)
```

```python
import base64
import pandas as pd
from PIL import Image
import matplotlib.pyplot as plt
from IPython.display import HTML
```

```python
def image_to_html(img_path):
    """Converts an image to an HTML img tag."""
    img = Image.open(img_path)
    buffered = BytesIO()
    img.save(buffered, format="PNG")
    img_str = base64.b64encode(buffered.getvalue()).decode('utf-8')
    return f'<img src="data:image/png;base64,{img_str}" width="150"/>'

def display_images_from_folder(folder_path):
    """Display all images from a folder in a DataFrame."""
    image_files = [f for f in os.listdir(folder_path) if f.lower().endswith(('png', 'jpg', 'jpeg', 'gif', 'bmp'))]
    
    df = pd.DataFrame(image_files, columns=["Image"])
    df["Image Path"] = df["Image"].apply(lambda x: os.path.join(folder_path, x))
    
    df["Image Display"] = df["Image Path"].apply(image_to_html)
    
    return HTML(df.to_html(escape=False, formatters={"Image Display": lambda x: x}))
```

```python
folder_path = "profile_images"
display(display_images_from_folder(folder_path))
```

## reCAPTCHA

Los reCAPTCHA son Captchas que además de evitar bots hacen machine learning, ya que se utilizan para la digitalización de libros, descifrando palabras que no  son legibles por computadores o para entrenar software de reconocimiento de imágenes. Por lo tanto, a diferencia de los Captchas que suelen ser al azar, los reCaptcha consisten en palabras específicas.
