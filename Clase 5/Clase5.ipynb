# Social Network Analysis

El análisis de redes sociales (SNA) es el uso de técnicas de la teoría de grafos para estudiar la estructura social y las relaciones humanas (afiliación, comunicación, amistad, odio.).

Ejemplos de estructuras sociales comúnmente visualizadas a través del análisis de redes sociales incluyen redes de medios sociales, difusión de memes, circulación de información, redes de amistad y conocidos, redes de aprendizaje entre pares, redes de negocios, redes de conocimiento, relaciones laborales, grafos de colaboración, parentesco y transmisión de enfermedades.

# Grafos

Conjunto de entidades llamados vértices o nodos unidos por enlaces llamados aristas o arcos, que permiten representar relaciones.

Matemáticamente, un grafo $G$ se denota como un par $G(V, E)$, donde $V$ representa el conjunto de nodos y $E$ representa el conjunto de aristas.

# Grado

- El número de aristas conectadas a un nodo es el grado de ese nodo. 

<div align="center">
    <img src="img/degree.png" alt="" width="10%" align="center" />
</div>

- En las redes sociales, el grado representa la cantidad de amigos que tiene un usuario determinado. 

### Grado en grafo no dirigido

El grado de un nodo $v_i$ a menudo se denota usando $d_i$. En cualquier grafo no dirigido, la suma de todos los grados de los nodos es igual al doble del número de aristas ($\sum_i d_i = 2 |E|$).

En Facebook, el grado representa el número de amigos del usuario y la persona de mayor grado es la más influyente.

### Grado en grafo dirigido

En el caso de los grafos dirigidos, los nodos tienen grados de entrada (aristas que apuntan hacia el nodo) y grados de salida (aristas que apuntan hacia afuera del nodo). Estos valores se denotan usando $d_{i}^{in}$ y $d_{i}^{out}$. En cualquier grafo dirigido, la suma de los grados de entrada es igual a la suma de los grados de salida ($\sum_i d_{i}^{in} = \sum_j d_{j}^{out}$). 

En Twitter, el grado de entrada es el número de seguidores y el de salida el número de cuentas que sigue la persona.

## Conectividad

Un nodo está conectado a otro si son adyacentes o existe una ruta que conecta a ambos. Un grafo es conexo si existe un camino entre cualquier par de nodos en él. Un **componente** en un grafo no dirigido es un subgrafo tal que existe un camino entre cada par de nodos dentro del componente.

```python
def plot_components(edges_g1, edges_g2):
    G1 = nx.Graph(edges_g1)
    G2 = nx.Graph(edges_g2)
    options = {"node_color": "white", "edgecolors": "black"}
    plantilla = "Grafo {}conexo ({} componentes)"
    fig = plt.figure(figsize=(13, 5))
    ax1 = fig.add_subplot(121) 
    ax2 = fig.add_subplot(122) 
    nx.draw(G1, **options, with_labels=True, ax=ax1)
    conex = "" if nx.is_connected(G1) else "no "
    ax1.set_title(plantilla.format(conex,nx.number_connected_components(G1)))
    nx.draw(G2, **options, with_labels=True, ax=ax2)
    conex = "" if nx.is_connected(G2) else "no "
    ax2.set_title(plantilla.format(conex,nx.number_connected_components(G2)))    
    plt.show()
```

```python
plot_components([(1, 2), (1, 3), (1, 4), (3, 4), (3, 5), (4, 5)], [(0, 1), (1, 2), (5, 6), (3, 4), (2, 0)])
```

## Camino más corto

Cuando un grafo está conectado, pueden existir múltiples caminos entre cualquier par de nodos. A menudo, estamos interesados en el camino que tiene la longitud más corta, este camino se llama el camino más corto.

```python
def plot_shortest_path(edges, n1, n2):
    G = nx.Graph(edges)
    options = {"node_color": "white", "edgecolors": "black"}
    plantilla = "El camino más corto entre {} y {} tiene tamaño {} y es {}.\nPromedio: {}"
    pos=nx.spring_layout(G)
    shortestpath = nx.shortest_path(G, source=n1, target=n2)
    sp = []
    for i in range(0,len(shortestpath)-1):
        sp.append((shortestpath[i],shortestpath[i+1]))
    nx.draw_networkx(G, pos, **options, with_labels=True)
    nx.draw_networkx_edges(G, pos, edge_color="black")    
    nx.draw_networkx_edges(G, pos, edgelist=sp, edge_color="red", width=2)
    plt.title(plantilla.format(n1,n2,len(shortestpath)-1,shortestpath, nx.average_shortest_path_length(G)))
    plt.show()
```

```python
plot_shortest_path([(1, 2), (1, 3), (1, 4), (3, 4), (3, 5), (4, 5)], 1, 5)
```

## Diámetro

El diámetro de un grafo se define como la longitud del camino más corto entre cualquier par de nodos en el grafo.

```python
def plot_diameter(edges):
    G = nx.Graph(edges)
    options = {"node_color": "white", "edgecolors": "black"}
    plantilla = "Diámetro: {}"
    nx.draw_networkx(G, **options, with_labels=True)
    plt.title(plantilla.format(nx.diameter(G)))
    plt.show()
```

```python
plot_diameter([(1, 2), (1, 3), (1, 4), (3, 4), (3, 5), (4, 5)])
```

## Degree Centrality

El grado de centralidad $C_d(v_i)$ para un nodo $v_i$ en un grafo no dirigido es $C_d(v_i)=d_i,$ donde $d_i$ es el grado del nodo.

En un grafo dirigido, se usan

- $C_d(v_i)=d_i^{in}$ (prestigio),
- $C_d(v_i)=d_i^{out}$ (sociabilidad),
- $C_d(v_i)=d_i^{in}+d_i^{out}$.

```python
def plot_degree_centrality(G):
    fig = plt.figure(figsize=(13, 5))
    ax1 = fig.add_subplot(131) 
    ax2 = fig.add_subplot(132) 
    ax3 = fig.add_subplot(133) 
    
    player = dict(nx.degree(G)).keys()
    degrees = dict(nx.degree(G)).values()
    in_degrees = dict(G.in_degree()).values()
    out_degrees = dict(G.out_degree()).values()
    
    degree = pd.DataFrame({'player':player, 'degrees':degrees, 'in_degrees':in_degrees, 'out_degrees': out_degrees})

    ordered = degree.sort_values(by = 'degrees')
    x_range = range(len(degree.index))
    y_range = range(math.ceil(max(degree.degrees))+1)
    ax1.barh(x_range, ordered['degrees'], color='#5162a4', edgecolor = 'black')
    ax1.set_yticks(x_range)
    ax1.set_yticklabels(ordered['player'])
    ax1.set_xticks(y_range)
    ax1.grid(axis = 'x', color = 'gray', alpha=0.5)
    ax1.set_ylabel("Player")
    ax1.set_xlabel("Degree (total number of passes played)")
    ax1.set_title("Successful passes (degrees) of each player")
    
    ordered = degree.sort_values(by = 'in_degrees')
    x_range = range(len(degree.index))
    y_range = range(math.ceil(max(degree.in_degrees))+1)
    ax2.barh(x_range, ordered['in_degrees'], color='#acd544', edgecolor = 'black')
    ax2.set_yticks(x_range)
    ax2.set_yticklabels(ordered['player'])
    ax2.set_xticks(y_range)
    ax2.grid(axis = 'x', color = 'gray', alpha=0.5)
    ax2.set_ylabel("Player")
    ax2.set_xlabel("in degree (total number of passes received)")
    ax2.set_title("Successful passes received (in degrees)")
    
    ordered = degree.sort_values(by = 'out_degrees')
    x_range = range(len(degree.index))
    y_range = range(math.ceil(max(degree.out_degrees))+1)
    ax3.barh(x_range, ordered['out_degrees'], color='#d6ab40', edgecolor = 'black')
    ax3.set_yticks(x_range)
    ax3.set_yticklabels(ordered['player'])
    ax3.set_xticks(y_range)
    ax3.grid(axis = 'x', color = 'gray', alpha=0.5)
    ax3.set_ylabel("Player")
    ax3.set_xlabel("Out Degree (Total Number of Passes Given)")
    ax3.set_title("Successful Passes Given (Out Degrees)")
            
    fig.tight_layout()
    plt.show()
```

```python
plot_degree_centrality(G)
```

## Clustering Coefficient

El coeficiente de clustering en un grafo mide cuántos amigos de un nodo también son amigos entre sí. En otras palabras, cuantifica la tendencia de los nodos a formar grupos interconectados, por ende, es la fracción de posibles triángulos que existen a través de un nodo:

$$ c_u = \frac{2T(u)}{deg(u)(deg(u)-1)}, $$

donde $T(u)$ es el número de triángulos a través del nodo $u$ y $deg(u)$ el grado del nodo $u$.

## Betweenness

Se puede traducir como “intermediación”, y mide si un nodo es intermediario en rutas más cortas entre otros nodos en una red. Indica en cierta manera si ese nodo es un punto de distribución entre comunidades. 

Se calcula contando cuántas veces un nodo está en el camino más corto entre otros pares de nodos y expresándolo como fracción de todos los posibles caminos.

<div align="center">
    <img src="img/betweenness_and_degree.png" alt="" width="20%" align="center" />
</div>

```python
def plot_centrality_metrics(G):
    fig = plt.figure(figsize=(13, 5))
    ax1 = fig.add_subplot(131) 
    ax2 = fig.add_subplot(132) 
    ax3 = fig.add_subplot(133) 
    
    player = dict(nx.pagerank(G, alpha=0.95)).keys()
    rank = dict(nx.pagerank(G, alpha=0.95)).values()
    cluster_coef = dict(nx.clustering(G)).values()
    betweenness = dict(nx.betweenness_centrality(G)).values()
    df = pd.DataFrame({'player':player, 'rank':rank, 'clustering_coef':cluster_coef, 'betweenness':betweenness})
    
    ordered = df.sort_values(by = 'rank')
    x_range = range(len(df.index))
    y_range = range(math.ceil(max(df['rank']))+1)
    ax1.barh(x_range, ordered['rank'], color='#5162a4', edgecolor = 'black')
    ax1.set_yticks(x_range)
    ax1.set_yticklabels(ordered['player'])
    ax1.set_xticks(y_range)
    ax1.grid(axis = 'x', color = 'gray', alpha=0.5)
    ax1.set_ylabel("Player")
    ax1.set_title("PageRank")

    ordered = df.sort_values(by = 'clustering_coef')
    x_range = range(len(df.index))
    y_range = range(math.ceil(max(df['clustering_coef']))+1)
    ax2.barh(x_range, ordered['clustering_coef'], color='#acd544', edgecolor = 'black')
    ax2.set_yticks(x_range)
    ax2.set_yticklabels(ordered['player'])
    ax2.set_xticks(y_range)
    ax2.grid(axis = 'x', color = 'gray', alpha=0.5)
    ax2.set_ylabel("Player")
    ax2.set_title("Clustering Coefficient")
    
    ordered = df.sort_values(by = 'betweenness')
    x_range = range(len(df.index))
    y_range = range(math.ceil(max(df['betweenness']))+1)
    ax3.barh(x_range, ordered['betweenness'], color='#d6ab40', edgecolor = 'black')
    ax3.set_yticks(x_range)
    ax3.set_yticklabels(ordered['player'])
    ax3.set_xticks(y_range)
    ax3.grid(axis = 'x', color = 'gray', alpha=0.5)
    ax3.set_ylabel("Player")
    ax3.set_title("Betweenness")    

    fig.tight_layout()
    plt.show()
```

```python
plot_centrality_metrics(G)
```

## Ruta de pase

```python
def edge_sort_key(path):
    return len(path), -sum(G[path[i]][path[i+1]]['weight'] for i in range(len(path)-1))
```

```python
def plot_passing_route(G, pos, start_node, end_node):
    plt.figure(figsize=(15, 8))    
    
    # Get all the simple paths from start to end
    all_paths = nx.all_simple_paths(G, source=start_node, target=end_node)

    # Sort paths based on the custom sorting function
    sorted_paths = sorted(all_paths, key=edge_sort_key)
    
    # Choose the path with the maximum weight while having the fewest edges
    shortest_path = sorted_paths[0]
    
    if len(shortest_path) == 2: route = [tuple(shortest_path)]  
    else: route = [(shortest_path[i],shortest_path[i+1]) for i in range(len(shortest_path)-1)]

    nx.draw_networkx_nodes(G, pos, node_size=2500, node_color='#eee9db')
    nx.draw_networkx_edges(G, pos, edge_color="black")    
    nx.draw_networkx_edges(G, pos, edgelist=route, 
                           edge_color="blue", width=3)
    labels = {node: node for node in G.nodes()}
    nx.draw_networkx_labels(G,pos,{n:player_number_dict[n].replace(' ','\n') for n in G.nodes()},font_size=10)
    plt.show()
```

```python
plot_passing_route(G, pos, 13, 9)
```

# Medidas de similaridad

Cuantifican la semejanza entre dos nodos en un grafo, aunque no existe una definición única de similitud.

En el contexto de una red de pases en un equipo de fútbol, una alta similitud entre dos nodos indica que estos jugadores exhiben patrones de pases similares.

## Coeficiente de Jaccard

El coeficiente de Jaccard entre el nodo $u$ y el nodo $v$ está definido como:

$$ J(u,v) = \frac{|N(u) \cap N(v)|}{|N(u) \cup N(v)|} $$

donde $N(u)$ denota el conjunto de vecinos de $u$.

```python
nodes = list(G.nodes())
similarity_matrix = np.zeros((len(nodes), len(nodes)))
for i, node1 in enumerate(nodes):
    for j, node2 in enumerate(nodes):
        successors1 = set(G.successors(node1))
        successors2 = set(G.successors(node2))
        intersection = successors1.intersection(successors2)
        union = successors1.union(successors2)
        if len(union) == 0:
            similarity = 0.0
        else:
            similarity = len(intersection) / len(union)
        similarity_matrix[i, j] = similarity
```

```python
players = [player_number_dict[key] for key in nodes]
```

```python
df_sim_jac = pd.DataFrame(similarity_matrix, index = players, columns = players)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df_sim_jac, annot=True, cmap='coolwarm') 
plt.title("Coeficiente de Jaccard")
plt.show()
```

## Similaridad Coseno

La similaridad coseno entre el nodo $u$ y el nodo $v$ está definido como:

$$ cos(u,v) = \frac{\sum_{i} N_i(u) \cdot N_i(v) }{\sqrt{\sum_{i} N_i(u)^2} \sqrt{\sum_{i} N_i(v)^2} } $$

donde $N(u)$ denota el conjunto de vecinos de $u$.


```python
nodes = list(G.nodes())
similarity_matrix = np.zeros((len(nodes), len(nodes)))
for i, node1 in enumerate(nodes):
    for j, node2 in enumerate(nodes):
        neighbors1 = set(G.successors(node1))
        neighbors2 = set(G.successors(node2))
        intersection = neighbors1.intersection(neighbors2)
        similarity = len(intersection) / (np.sqrt(len(neighbors1)) * np.sqrt(len(neighbors2)))
        similarity_matrix[i, j] = similarity
```

```python
df_sim_cos = pd.DataFrame(similarity_matrix, index = players, columns = players)
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df_sim_cos, annot=True, cmap='coolwarm')
plt.title("Similaridad Coseno")
plt.show()
```
