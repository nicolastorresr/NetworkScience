# Módulo requests

El módulo `requests` se utiliza para realizar solicitudes HTTP a una URL específica y devuelve la respuesta. Las solicitudes de Python proporcionan funcionalidades integradas para administrar tanto la solicitud como la respuesta.

Una solicitud HTTP está destinada a recuperar datos de un URI específico o enviar datos a un servidor. Funciona como un protocolo de solicitud-respuesta entre un cliente y un servidor.

```python
# pip install requests
```

```python
import requests
```

```python
r = requests.get('https://usm.cl/')
```

```python
r.status_code
```

- 200 OK – La solicitud fue exitosa.
- 400 Bad Request – La solicitud tiene un formato incorrecto o no es válida.
- 401 Unauthorized – La solicitud requiere autenticación.
- 403 Forbidden – El servidor se negó a cumplir con la solicitud.
- 404 Not Found – No se pudo encontrar el recurso solicitado.
- 500 Internal Server Error – El servidor encontró una condición inesperada.

```python
# contenido de la respuesta en Bytes
r.content[:1000]
```

```python
# contenido de la respuesta en Unicode
r.text[:1000]
```

El método `find_all()` examina los descendientes de una etiqueta y recupera todos los descendientes que coinciden con sus filtros.

```python
usm_sitios = []
for link in soup.find_all('a'):
    usm_sitios.append(link.get('href'))
```

# Rastreando imágenes

```python
images_list = []
imagenes = []
images = soup.find_all('img')
for image in images:
    src = image.get('src')
    if re.search("\.(jpg|gif|png|bmp)$", src):
        imagenes.append(src)
    alt = image.get('alt')
    images_list.append({"src": src, "alt": alt})
```

```python
import matplotlib.pyplot as plt
from urllib.parse import quote
from PIL import Image
import requests
from io import BytesIO
```

```python
def plot_images(images):
    n_rows = (len(images) - 1) // 5 + 1
    plt.figure(figsize=(15, n_rows * 3))
    for i, url in enumerate(images):
        encoded_url = quote(url, safe=':/')
        try:
            # Usamos requests con un User-Agent para evitar el error 403
            headers = {'User-Agent': 'Mozilla/5.0'}
            response = requests.get(encoded_url, headers=headers)
            response.raise_for_status()
            img = Image.open(BytesIO(response.content))
            plt.subplot(n_rows, 5, i + 1)
            plt.imshow(img)
            plt.axis('off')
        except Exception as e:
            print(f"Error loading image {url}: {e}")
    plt.subplots_adjust(wspace=0.1, hspace=0.5)
```

```python
plot_images(imagenes)
```

## Guardar datos en un CSV

```python
import csv
```

```python
filename = 'imagenes.csv'
with open(filename, 'w', newline='') as f:
    w = csv.DictWriter(f,['alt','src'])
    w.writeheader()     
    w.writerows(images_list)
```

## Recopilar las imágenes asociadas

```python
page = browser.get_current_page()
all_images = page.find_all('img')
```

```python
image_source = []
for image in all_images:
    image = image.get('src')
    image_source.append(image)
image_source = [image for image in image_source if image.startswith('https')]
```

```python
def plot(images):
    n_rows = (len(images) - 1) // 5 + 1
    plt.figure(figsize=(15, n_rows * 3))
    for i in range(len(images)):
        plt.subplot(n_rows, 5, i + 1)
        url = images[i]
        try:
            response = requests.get(url, headers={'User-Agent': 'Mozilla/5.0'})
            img = Image.open(BytesIO(response.content))
            plt.imshow(img)
            plt.axis('off')
        except Exception as e:
            print(f"Error loading image {url}: {e}")
    plt.subplots_adjust(wspace=0.1, hspace=0.5)
    plt.show()
```

```python
plot(image_source)
```
