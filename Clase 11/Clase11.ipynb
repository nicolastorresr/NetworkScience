# Fake News

Las “noticias falsas” son información fabricada que imita el contenido de los medios de noticias en la forma, pero no en el proceso organizacional o la intención. Los medios de noticias falsas, a su vez, carecen de las normas y procesos editoriales de los medios de comunicación para garantizar la exactitud y credibilidad de la información.

# Sesgos Cognitivos y Fake News

## Sesgo de Confirmación

Tendencia a buscar, interpretar y recordar la información de manera que confirme nuestras creencias previas, ignorando o descartando aquella que las contradiga. Aumenta la difusión selectiva: compartimos más lo que coincide con nuestra visión del mundo.

Ejemplo: Una persona que desconfía de las vacunas tenderá a leer, compartir y creer artículos falsos que refuercen su desconfianza, incluso si la evidencia científica demuestra lo contrario.

```python
import networkx as nx
import matplotlib.pyplot as plt
import random
from ipywidgets import interact
```

```python
# Crear red social
G = nx.erdos_renyi_graph(20, 0.15)

# Asignar ideología (0=azul, 1=rojo)
ideology = {n: random.choice([0,1]) for n in G.nodes()}
nx.set_node_attributes(G, ideology, "ideology")

# Noticia con ideología "roja"
news_ideology = 1  
spread = {n: False for n in G.nodes()}

# Semilla inicial
seed = random.choice(list(G.nodes()))
spread[seed] = True
```

```python
# Guardamos estados paso a paso
steps = 5
states = [spread.copy()]
for t in range(steps):
    new_spread = states[-1].copy()
    for node in G.nodes():
        if states[-1][node]:
            for neigh in G.neighbors(node):
                if not states[-1][neigh] and ideology[neigh] == news_ideology:
                    if random.random()<0.5:  
                        new_spread[neigh] = True
    states.append(new_spread)

# Layout fijo de nodos
pos = nx.spring_layout(G, seed=42)
```

```python
# Función para dibujar en un paso dado
def plot_step(t=0):
    spread = states[t]
    colors = []
    for n in G.nodes():
        if spread[n]:
            colors.append("red" if ideology[n]==1 else "blue")
        else:
            colors.append("pink" if ideology[n]==1 else "lightblue")
    
    fig, ax = plt.subplots(figsize=(6,6))
    nx.draw(G, pos=pos, node_color=colors, with_labels=True, ax=ax)
    ax.set_title(f"Propagación en paso {t} (Sesgo de Confirmación)")
    plt.show()
```

```python
interact(plot_step, t=(0, steps));
```

## Heurística de Disponibilidad

Mecanismo mental por el cual juzgamos la probabilidad o relevancia de un hecho en función de lo fácilmente que recordamos ejemplos similares. Las noticias impactantes o emocionales (violencia, corrupción, teorías conspirativas) se recuerdan más fácilmente → parecen más comunes. Esto explica por qué el contenido sensacionalista o alarmista se viraliza más que el informativo y balanceado.

Ejemplo: Tras ver varias noticias (reales o falsas) sobre ataques con tiburones, las personas creen que son muy frecuentes, aunque estadísticamente sean extremadamente raros.

## Polarización Afectiva

Tendencia a ver a quienes piensan distinto no solo como diferentes, sino como enemigos o moralmente inferiores. Facilita la propagación de contenidos que atacan a “los otros”, ya que generan emociones fuertes (enojo, odio, desprecio). Aumenta la hostilidad digital y el hate speech.

Ejemplo: En política, no basta con estar en desacuerdo con el “otro partido”; se les percibe como corruptos, ignorantes o peligrosos.

```python
# Red con dos comunidades (dos cliques de 10 nodos)
G = nx.connected_caveman_graph(2, 10)

# Grupo A = rojo, Grupo B = azul
group = {n: 0 if n < 10 else 1 for n in G.nodes()}
nx.set_node_attributes(G, group, "group")

# Fake news: ataque contra Grupo Azul
target_group = 1
spread = {n: False for n in G.nodes()}
spread[0] = True  # Semilla en Grupo Rojo

# Guardamos la evolución paso a paso
steps = 6
history = []

current_spread = spread.copy()
history.append(current_spread.copy())
```

```python
for t in range(steps - 1):
    new_spread = current_spread.copy()
    for node in G.nodes():
        if current_spread[node]:
            for neigh in G.neighbors(node):
                if not current_spread[neigh] and group[neigh] == target_group:
                    # No comparten noticias que ataquen a su propio grupo
                    continue
                if not current_spread[neigh]:
                    new_spread[neigh] = True
    current_spread = new_spread
    history.append(current_spread.copy())

# Layout fijo para que no cambie cada vez
pos = nx.spring_layout(G, seed=42)
```

```python
# Función de dibujo para interact
def plot_step(step=0):
    spread = history[step]
    colors = []
    for n in G.nodes():
        if spread[n]:
            colors.append("red" if group[n] == 0 else "blue")
        else:
            colors.append("pink" if group[n] == 0 else "lightblue")

    fig, ax = plt.subplots(figsize=(6, 6))
    nx.draw(G, pos, node_color=colors, with_labels=True, edge_color="gray", ax=ax)
    ax.set_title(f"Polarización Afectiva - Paso {step}")
    plt.show()
```

```python
interact(plot_step, step=(0, steps-1));
```

# Modelos de Detección Temprana de Fake News (II)

## Enfoques basados en la red social y comportamiento

- **Dinámica de propagación:** Fake news tienden a difundirse más rápido y con mayor profundidad que las reales. Detección temprana al observar patrones de cascada en red.

- **Características de usuarios y bots:** Bots generan actividad anómala (muchos posts en poco tiempo, horarios 24/7).

- **Métodos híbridos:** Combinan análisis de contenido + patrones de difusión. Ejemplo: Modelos de grafos (Graph Neural Networks – GNNs) aprenden la estructura de la propagación para clasificar rumores en tiempo real.

# 1- Conceptos básicos de los grafos 

Antes de saltar a los mecanismos de las redes neuronales de grafos, comenzaremos recordando que los grafos son representaciones de datos no estructurados, permiten tamaños y formas no fijas y se utilizan para simular datos de escenarios complejos del mundo real donde las estructuras regulares como imágenes, audio o texto secuencial pueden fallar.

# 2- Graph Neural Networks (GNNs)

Las redes neuronales de grafos (GNN) son un tipo especial de arquitectura de red neuronal diseñada específicamente para trabajar con grafos. Sin embargo, los grafos no son imágenes ni texto, entonces deberíamos idear algo diferente a las CNN o RNN. Por ejemplo, los grafos son invariantes al orden (el orden de los nodos no es relevante). Con las imágenes tenemos el comportamiento opuesto: si procesamos un píxel en un orden diferente al de su posición original, la imagen ya no se verá igual.

# 4- El Marco General de las GNN

El proceso de compartir datos entre nodos se realiza en dos pasos conocidos como Agregación y Combinación. Durante el primer paso, la información de todos los nodos que comparten una arista con el nodo central se recopila y transforma para crear una representación del vecindario. Luego, en el segundo paso, el resultado de esta agregación se combina con el valor actual del nodo para producir su nuevo valor.

# Graph Convolutional Networks (GCNs)

## CNN

En las imágenes, los píxeles se ordenan estructuralmente en una cuadrícula y los filtros en la operación de convolución se deslizan sobre la imagen con un tamaño de desplazamiento predeterminado. La salida entrega un tamaño más pequeño que la imagen de entrada, pero tiene una vista de mayor nivel.

<div align="center">
    <img src="img/image_convolution.png" alt="" width="50%" align="center" />
</div>

## GCN

En los grafos, los nodos se ordenan de manera no estructural y el tamaño de la vecindad difiere entre los nodos. La convolución en grafos toma el promedio de las características de un nodo determinado y sus vecinos para calcular una representación actualizada de los valores del nodo. A través de esta operación de convolución, la representación del nodo captura información local de grafo.

<div align="center">
    <img src="img/graph_convolution.png" alt="" width="50%" align="center" />
</div>

## Convolución en grafos

La operación convolución promedia las características de los nodos vecinos junto con las del nodo de destino. Luego se multiplica con un vector de peso (W) y su salida actualiza las características del nodo de destino (los valores de nodo actualizados también se denominan node embeddings).

<div align="center">
    <img src="img/convolution_in_graphs.png" alt="" width="60%" align="center" />
</div>

## Niveles de abstracción

El número de convoluciones en el grafo determina cuántos vecinos y cuánta información se agrega en las nuevas características del nodo.
